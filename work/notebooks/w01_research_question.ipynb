I am choosing Lane 2 because it directly addresses limited human editorial capacity by framing content prioritization as a decision-support problem. Rather than relying on rigid rules, this lane allows us to learn observable decay patterns from search and analytics data to rank which mature pages require human review first.

* **Research Question:** Which mature, previously visible pages show significant traffic decay and should be prioritized for content refresh or optimization?
* **Unit of Analysis:** A single pseudonymized content item (page).
* **Target Output:** A ranked queue of candidate pages with predicted decline risk probabilities, baseline priority scores, and reason codes.
* **Decision & Action:** Editorial and SEO teams use the ranked queue to allocate limited review and rewrite hours toward high-impact pages instead of guessing.
* **Cost of a Wrong Call:**
  * *False Positive:* Wastes reviewer and writer hours on pages that do not need updating.
  * *False Negative:* High-value pages quietly bleed search traffic and rankings without being noticed.
* **Why ML Helps:** Simple heuristic thresholds fail to handle trade-offs between conflicting signals (e.g., strong impressions vs. decaying CTR vs. content age). A learned model captures non-linear decay interactions across multiple features.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
```

Based on the starter dataset (`data/raw/content_refresh_anonymized.csv`), three key baseline numbers justify this lane:

1. **Large Decay Population:** Out of 30,000 mature eligible pages (age >= 90 days, impressions > 0), **16,262 pages (54.2%)** exhibit a downward trend (`trend_direction == 'down'`).
2. **Heuristic Inefficiency:** A standard strict baseline rule (update age >= 180 days and impressions >= 500) captures only **17 pages (0.1%)**, missing thousands of decaying pages.
3. **Severe Demand Skew:** Impressions are heavily right-skewed—the median 90-day impressions is **731**, while the 75th percentile reaches **3,615**, demonstrating why opportunity ranking must balance decay probability with traffic magnitude.

```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")
df_filtered = df[(df["impressions_90d"] > 0) & (df["content_age_days"] >= 90)].copy()

total_pages = len(df_filtered)
declining_count = (df_filtered["trend_direction"] == "down").sum()
decline_pct = (declining_count / total_pages) * 100

stale_visible = df_filtered[
    (df_filtered["days_since_last_update"] >= 180) & (df_filtered["impressions_90d"] >= 500)
]
stale_visible_pct = (len(stale_visible) / total_pages) * 100

print(f"Total eligible pages: {total_pages:,}")
print(f"Pages in decline: {declining_count:,} ({decline_pct:.1f}%)")
print(f"Stale visible pages: {len(stale_visible):,} ({stale_visible_pct:.1f}%)")
```
